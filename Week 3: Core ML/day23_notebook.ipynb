# Concept

As the number of features increases, data becomes sparse and models become harder to train.

PCA (Principal Component Analysis) reduces the number of features while preserving as much information as possible.

Instead of keeping the original features, PCA creates new features called Principal Components.


# Real World Example

Netflix, Spotify, and recommendation systems often work with thousands of features.

Reducing dimensions helps:
- Reduce training time
- Remove noise
- Improve visualization
- Reduce storage requirements


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

sns.set_theme()
```

## Load Dataset

```
wine = load_wine()

X = wine.data
y = wine.target

print("Original Shape:", X.shape)
```

## Standardize Features

```
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## Apply PCA

```
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

print("Reduced Shape:", X_pca.shape)
```

## Explained Variance

```
print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)

print("\nTotal Variance Retained:")
print(sum(pca.explained_variance_ratio_))
```

## Visualize Principal Components

```
plt.figure(figsize=(8,6))

scatter = plt.scatter(
    X_pca[:,0],
    X_pca[:,1],
    c=y
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Visualization")
plt.show()
```

## Scree Plot

```
pca_full = PCA()
pca_full.fit(X_scaled)

plt.figure(figsize=(8,5))
plt.plot(
    np.cumsum(pca_full.explained_variance_ratio_),
    marker='o'
)

plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Explained Variance by Components")
plt.show()
```

# Interview Question

Q: Why do we scale features before PCA?

Answer:

PCA relies on variance.

Features with larger scales dominate variance calculations and can distort the principal components.

Scaling ensures all features contribute fairly.


# ML Spotlight

PCA is often used before visualization, clustering, anomaly detection, and compression.

Even modern deep learning pipelines frequently use PCA for exploratory data analysis and feature inspection.


# Key Takeaways

✅ PCA reduces dimensions while preserving information
