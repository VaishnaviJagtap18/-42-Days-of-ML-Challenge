# Day 03: Descriptive Statistics

Mean, Median, Mode, Variance, Standard Deviation, Skewness, Kurtosis, and Log Transformation.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

pd.set_option('display.max_columns', None)
sns.set_theme()
```

## Load Dataset

```
df = pd.read_csv('train.csv')
print(df.shape)
df.head()
```

## Pandas Describe

```
df.describe().T
```

## Mean, Median, Mode, Variance, Std, Skewness, Kurtosis

```
sale_price = df['SalePrice']

print('Mean:', np.mean(sale_price))
print('Median:', np.median(sale_price))
print('Mode:', stats.mode(sale_price, keepdims=True))
print('Variance:', np.var(sale_price))
print('Std Dev:', np.std(sale_price))
print('Skewness:', stats.skew(sale_price))
print('Kurtosis:', stats.kurtosis(sale_price))
```

## SciPy Describe

```
stats.describe(sale_price)
```

## Identify Skewed Columns

```
num_cols = df.select_dtypes(include=np.number).columns
skew_vals = df[num_cols].skew().sort_values(ascending=False)
skewed_cols = skew_vals[abs(skew_vals) > 1]
skewed_cols
```

## SalePrice Distribution

```
sns.histplot(df['SalePrice'], kde=True)
plt.title('Original SalePrice Distribution')
plt.show()
```

## Log Transformation

```
df['LogSalePrice'] = np.log1p(df['SalePrice'])

print('Original Skewness:', df['SalePrice'].skew())
print('Log Skewness:', df['LogSalePrice'].skew())
```

## Before vs After

```
plt.figure(figsize=(12,4))

plt.subplot(1,2,1)
sns.histplot(df['SalePrice'], kde=True)

plt.subplot(1,2,2)
sns.histplot(df['LogSalePrice'], kde=True)

plt.show()
```
